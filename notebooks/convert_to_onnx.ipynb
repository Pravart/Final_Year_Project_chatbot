```python
from pathlib import Path
from optimum.onnxruntime import ORTModelForSequenceClassification
from transformers import AutoTokenizer

# Local model path
model_path = Path("../models/distilbert_emotion").resolve()

# Output folder
onnx_path = Path("../models/distilbert_emotion_onnx")

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(str(model_path))

# Export model to ONNX
model = ORTModelForSequenceClassification.from_pretrained(
    str(model_path),
    export=True,
    local_files_only=True
)

# Save exported model
model.save_pretrained(str(onnx_path))
tokenizer.save_pretrained(str(onnx_path))

print("✅ ONNX model saved to:", onnx_path)
```

```python
from pathlib import Path
from optimum.onnxruntime import ORTModelForSequenceClassification
from transformers import AutoTokenizer

# Local sub-emotion model
model_path = Path("../models/distilbert_sub_emotion").resolve()

# Output folder
onnx_path = Path("../models/distilbert_sub_emotion_onnx")

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(str(model_path))

# Export model to ONNX
model = ORTModelForSequenceClassification.from_pretrained(
    str(model_path),
    export=True,
    local_files_only=True
)

# Save exported model
model.save_pretrained(str(onnx_path))
tokenizer.save_pretrained(str(onnx_path))

print("✅ Sub-emotion ONNX model saved to:", onnx_path)
```

```python
from pathlib import Path
from onnxruntime.quantization import quantize_dynamic, QuantType

model_fp32 = Path("../models/distilbert_emotion_onnx/model.onnx")
model_int8 = Path("../models/distilbert_emotion_onnx/model_int8.onnx")

quantize_dynamic(
    model_input=str(model_fp32),
    model_output=str(model_int8),
    weight_type=QuantType.QInt8
)

print("✅ Main emotion model quantized successfully!")
```
